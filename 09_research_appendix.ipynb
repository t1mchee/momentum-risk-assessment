# Research appendix — methods and results

This compact appendix supports the [main assessment](09_economic_vulnerability.html).
I record the methods, checks and results of the research below. The main calculations
use volatility-scaled historical returns and rolling-beta scenarios, with historical
and GARCH path estimates shown separately. Returns are fixed-notional percentage points.

**Section references:** baseline probability/calibration → E01/E03/E08; conditional
response → E02/E12; data quality → E09; constituents → E14–E15; options/text example
→ E09/E16; path risk → E18–E19/E21; timing/state approaches → E17–E18;
recent validation → E20.

The [optional complete archive](SUPPLEMENTARY_RESEARCH_ARCHIVE.html)
contains all saved executed figures, full tables and expandable calculation code.
It is **not required reading**. Underlying machine-
readable score files remain in the reproduction package, not as full inline tables.

**Evaluation samples:** chronological evaluation is not untouched validation after
model selection. Some reserved observations extending into 2026 were later examined
during development. The worked assessment is dated 2022; E20 separately evaluates
the unchanged specifications on 2023–July 2026. Broader-sample comparisons are labelled.
The 2026 options example is separate. Pointwise intervals do not correct for the
earlier model search. E20 extends validation rather than selecting another model family. Negative tests apply to the chosen specifications,
not every conceivable model. Human PM usefulness and integrated AI text performance
are not established by these numerical studies.

## E01. Initial sequential probability selection

**Method:** I fitted probability models chronologically, defined the loss threshold in the initial training sample and tuned models within development samples.

**Checks:** I checked event frequencies, probability scores and whether training observations preceded each forecast.

**Result:** The added market-state variables did not consistently improve the volatility model. The originally reserved observations were later examined during development.

The threshold is the fifth percentile of 522 weekly sums in 1964–1973: −2.157 pp, rounded to −2.16 pp for subsequent tests.

<a id="e02"></a>
## E02. Market-state conditional scenarios

**Method:** I fitted separate long- and short-leg regressions using market returns and the preceding market state, including different rising- and falling-market slopes.

**Checks:** I compared annual errors, rebound observations and results after excluding crisis periods.

**Result:** The estimates describe greater loser-leg sensitivity in some rebounds. Prediction errors and low coverage of outcome bands limit confidence in scenario magnitudes.

<a id="e03"></a>
## E03. Literature model and positioning comparison

**Method:** I compared seventeen probability and sixteen downside-quantile forecasts, including volatility, inferred-state and direct-tail models, and a lagged CFTC positioning measure.

**Checks:** I used common targets and dates, paired error comparisons and the reported model-family search adjustment.

**Result:** I used EWMA-scaled historical returns to obtain probabilities and quantiles from the same distribution. The tests do not establish a probability advantage after the model-family adjustment. The positioning measure concerns a different exposure.

The positioning series is CFTC consolidated S&P 500 futures leveraged-money (long−short)/open interest and its four-report change, delayed fourteen or twenty-eight days. Fits start in 2012 using history from 2006. The family adjustment uses shared bootstrap draws and the 95th percentile of the maximum absolute centred, standardised error difference (max-t). It adjusts that comparison family, not the complete selection after trying many specifications.

## E04. Integrated distribution and portfolio construction tests

**Method:** I tested weekly skew-t GJR-GARCH, GARCH-scaled historical returns and separate large negative, large positive and volatility-standardised return targets.

**Checks:** I checked past-only inputs, perturbed future outcomes to test for leakage, and checked quantile ordering and probabilities.

**Result:** The state variables did not improve the selected forecasts. Predicting changes in volatility did not establish improved prediction of distribution shape in these tests.

For these tests, $Y_w$ is summed weekly factor P&L and
$z_w=Y_w/(\sqrt5\sigma_t)$ uses EWMA scale at the preceding cutoff.
The four events are $Y_w<-2.16$, $Y_w>2.16$, $z_w<-1.5$ and $z_w>1.5$.
The scale logit uses log EWMA. The state logit adds log 126-day market volatility,
the bear indicator, trailing 252-day factor beta, trailing 21-day compounded market
return, and the three bear interactions listed below.
Both use training-standardised inputs and L2 logistic regression with C=1,
refitted annually on expanding completed weeks from 1974. The state feature list is
log scale, log market volatility, bear, beta, rally, bear×log market volatility,
bear×beta and bear×rally. These are the definitions behind the four rows.

| Four-target falsification | Scale log loss | State log loss |
| --- | ---: | ---: |
| Absolute left | .26428 | .27480 |
| Absolute right | .30841 | .31289 |
| Standardised negative | .26234 | .26682 |
| Standardised positive | .34079 | .34145 |

On the broader 2,221-week sample, EWMA/FHS probability log loss is .26376,
weekly skew-t GJR-GARCH .26510, GARCH-t/FHS .26511. Point scores do not establish
general superiority; samples differ from the through-2022 headline comparison.

<a id="e05"></a>
## E05. Calibration, onset, parameter and economic checks

**Method:** I tested recalibration, first-loss alerts, interim losses, nineteen parameter settings, crisis exclusions and transfer between small- and large-stock portfolios.

**Checks:** I compared calibration, first and continuing loss weeks, parameter settings and results after removing major crises.

**Result:** The model underpredicts losses in bear states, and its results vary with parameters and episodes. The first-loss tests did not establish a reliable reversal warning.

<a id="e06"></a>
## E06. PM scenarios and change accounting

**Method:** I calculated dated conditional leg responses, counted similar historical weeks and separated changes in estimated scale from changes in the historical sample.

**Checks:** I checked that contributions added to the total and examined sensitivity to lookback and the amount of similar history.

**Result:** These calculations describe how the estimates changed. I did not test whether the added information improved PM decisions.

<a id="e07"></a>
## E07. Conditional quantiles and joint tails

**Method:** I fitted conditional quantile regressions and measured both legs on the same adverse net-return windows.

**Checks:** I measured quantile coverage and pinball loss and checked for crossing quantiles and simultaneous leg breaches.

**Result:** The quantile bands were breached too often in bear/rebound observations. I kept the results as a comparison rather than using them as the main scenario output.

<a id="e08"></a>
## E08. Baseline validation and distribution appendix

**Method:** I compared the historical-return model with a logit using the same EWMA input and examined calibration through time, loss runs and leg contributions.

**Checks:** I checked 2,035 evaluation weeks, 132 dated assessments, the same-EWMA comparison and the complete loss-run record.

**Result:** The probability-score difference is uncertain. Historical leg contributions and interim losses describe additional aspects of the returns, but do not establish improved forecasts.

| Baseline check | Result |
| --- | --- |
| Retained minus matched-EWMA Brier | −.00058; paired interval approximately [−.00353,+.00270] |
| Retained bear-state average probability | 16.61% versus 24.37% observed |
| Same-severity illustration | Roughly −2.04/−2.08 pp tails conceal short-leg versus long-leg losses |

The main submission embeds the construction diagram and through-time calibration.
Distribution coherence is a rationale for retaining the reference—not a claim
that it is the most conservative or best-calibrated stress probability.

## E09. Return distributions and company-data audit

**Method:** I examined return distributions over 1, 5, 10 and 20 trading days and audited company-price data for duplicate identifiers, missing observations and corporate actions.

**Checks:** I varied lookbacks and non-overlapping window offsets, removed the largest shock and checked corporate actions against sources.

**Result:** Verified splits materially changed apparent losses. The remaining data problems prevented a reliable reconstruction of the public factor. Central-quantile changes did not always coincide with worse tail losses.

**The two stock constructions differ.** The corporate-action correlation uses a
monthly reconstructed portfolio over February 2014–December 2022. It ranks split-adjusted
price changes from twelve months before formation to one month before formation,
requires at least 95% formation-price coverage and 500 eligible names, and selects
the highest/lowest floor(n/10) stocks. Weights are equal within each leg for the quoted
correlation. Membership and weights are fixed for each holding month. On each usable
day, at least 95% of each leg's weight must have observed returns; weights are then
renormalised within the observed subset. The 1,491 observations are matched usable
daily returns, not a continuous 1,491-day period. The saved audit reports the count
and eligible months rather than a separate exact-date ledger.

The later PCA work reuses cached month-end winner/loser membership. The available upstream
code takes the last 273 price rows through formation, calculates returns within that
slice and retains the first 252 rows, excluding the final 21 rows. The first retained
return is missing, leaving at most 251 observed returns. This describes the available
code; I have not independently established the version that generated the cache. It requires
80% observed formation returns, screens absolute daily moves above 60%, and ranks
eligible IWV members at the formation date into deciles. Missing screened formation returns are
treated as zero within the product. I did not regenerate this cached membership in
the submission; that convention is a limitation of the inherited proxy.
The PCA weights are proportional to each stock's reported price times the number
of its shares held by IWV, rather
than the equal weights used for the audit correlation.

I attempted 101 cached portfolio dates from 31 January 2014 to 31 December 2022;
36 passed. A usable snapshot is at most seven calendar days old and has positive
formation values for at least 90% of each original leg's names. The preceding
252-trading-day calendar must have at least one observed stock return on 90% of dates.
Retained names must have complete returns on that observed subset of dates,
nonzero variation, no absolute return above 45%,
and no unresolved inferred split. Each retained leg needs thirty names and 80% of
its known formation value. I renormalise each accepted leg to unit absolute weight.

Consecutive-date variance comparisons use daily sample covariances over each date's
preceding 252-day window, restricted to names complete at both dates. Each of the
four date/leg combinations must retain 95% of accepted weight; remaining weights
are renormalised separately. In the PCA offset formula, L and S denote only returns
represented by the first three components. I standardise each stock, project onto
the eigenvectors, then multiply loadings by stock return standard deviations and
signed weights to recover pp units. Market removal is an OLS regression of each
stock on an intercept and market return over the same window, before PCA.

<a id="e10"></a>
## E10. Distribution panel, decomposition and dependence

**Method:** I decomposed historical tail losses into leg contributions, groups of loss windows and changes as observations entered or left the sample.

**Checks:** I checked fractional tail weights, additive contributions and sensitivity to overlapping windows and individual episodes.

**Result:** The components describe changes in the observed sample. I did not find a stable regime classification that justified an additional forecasting input.

<a id="e11"></a>
## E11. Final distribution and path checks

**Method:** I reproduced assessments at scheduled historical dates, compared interim and endpoint losses, and checked the effect of observations leaving rolling windows.

**Checks:** I checked that later observations did not alter earlier calculations and that interim and endpoint contributions reconciled.

**Result:** Some interim losses had recovered by the endpoint. Some changes in trailing estimates arose because old observations left the window.

<a id="e12"></a>
## E12. Exposure, shape and macro forecasting / conditional response

**Method:** I tested conditional exposure regressions and additional exposure, distribution-shape and macro inputs, using 126-, 252- and 504-trading-day exposure estimates.

**Checks:** I compared a volatility model with the same number of regressors, annual and subset errors, and 26/52-week block intervals.

**Result:** Rolling beta had slightly lower pooled error than the more complex conditional model. The additional predictors did not consistently improve unconditional loss forecasts.

| Conditional reference/challenger | Pooled MAE | Pooled MSE |
| --- | ---: | ---: |
| Selected rolling beta | 1.3126 | 3.8277 |
| Nonlinear exposure challenger | 1.3176 | 3.8355 |
| Volatility-aware comparator | 1.3606 | — |

Only 19/39 years favour the nonlinear challenger over the volatility comparator
on MAE; recent and crisis-excluded intervals include zero. A 504-session exposure
window loses its pooled advantage. Supplied market returns test a conditional
relationship, not market-move prediction.

<a id="e13"></a>
## E13. Macro scenario follow-on

**Method:** I added market-rate and credit-spread changes to rolling conditional leg regressions and varied the assumed data-availability lag.

**Checks:** I used common samples, varied the availability lag and compared paired net and leg errors.

**Result:** The additions did not consistently reduce net or leg errors. The macro histories are current-vintage data rather than a complete archive of historical releases.

<a id="e14"></a>
## E14. Constituent PCA

**Method:** I fitted PCA to the reconstructed stock portfolios, repeated it after removing market and sector returns, and compared correlation- and covariance-based estimates.

**Checks:** I checked data coverage, component sign and rotation sensitivity, variance identities and bootstrap variation.

**Result:** The first component share largely tracked average correlation. Only two portfolio dates passed the subsequent-month coverage check, so I did not use PCA as a forecast input.

The stock formation, coverage, weighting and PCA calculations are defined in [E09](#e09). The PCA analysis uses the cached membership and ETF-value weights described there, not the equal-weight audit portfolio.

<a id="e15"></a>
## E15. Book changes, tail concentration and common exposure

**Method:** I separated changes in portfolio variance into weight and covariance effects and measured stock/sector loss concentration and covariance between the legs.

**Checks:** I checked coverage of both legs at both dates, contribution totals, episode and stock omissions, and closely spaced PCA components.

**Result:** The coverage checks left 36 portfolio snapshots and 23 consecutive-date comparisons. The results concern those approximate portfolios, whose holdings differ from the public factor.

<a id="e16"></a>
## E16. MTUM options audit and pricing proof of concept

**Method:** I audited seventeen MTUM options files for dates, spreads, contract coverage and price consistency, then calculated an illustrative put-spread payoff and quoted cost.

**Checks:** I varied the bid–ask spread screen and checked timestamps, displayed sizes and put-spread arithmetic.

**Result:** The files contained seven distinct underlying trade dates. Too few out-of-the-money quotes passed the spread screens to estimate a downside distribution reliably.

## E17. Recovery hazard specifications

**Method:** I fitted recovery-hazard models for 2.5, 5 and 10 pp drawdowns, using volatility, distance to recovery, time elapsed and exposure.

**Checks:** I checked which observations were still awaiting recovery, chronological fits, synthetic examples and sensitivity to the longest episode.

**Result:** Distance to recovery explained more than elapsed time in these specifications. Defining recovery as a return to the old peak combined 2008–2022 into one long episode; the primary evaluation contained only fourteen recoveries.

The E17 table uses the 5 pp entry threshold. I accumulated weekly fixed-notional
returns from 1974, recorded the running peak, and opened an episode when the distance
below that peak reached 5 pp. The event is recovery to the peak by the next weekly
cutoff. Rows are active episode-weeks, not separate episodes. Evaluated cutoffs run
from 5 September 1986 to 23 December 2022; early years lack sufficient training events.
Fits use completed active weeks before the year's first forecast, at least 100 rows
and ten recoveries. Unrecovered episodes remain censored at the end of the data.

| Model label | Inputs |
| --- | --- |
| rate | Training recovery frequency |
| vol | log weekly EWMA scale |
| depth | vol plus log(1 + distance below peak in pp) |
| age | depth plus log(1 + weeks since entry) |
| exposure | depth plus net beta and rising/falling slope difference |
| full | exposure plus log age |
| distance | log(1 + distance below peak / weekly EWMA scale) |
| distance_age / distance_exposure / distance_full | distance plus age, exposure, or both |

The logits use standardised training inputs and C=.1. Lower Brier/log loss indicates
lower probability error; n counts active weeks and events counts next-week recoveries.

## E18. Local stress transitions and first passage

**Method:** I tested transitions into and out of rolling stress states against a comparator using the current score and the return about to leave the window. I also estimated twenty-day loss-path probabilities.

**Checks:** I excluded unfinished outcome windows from training and varied block lengths, historical-path sample lengths and barrier definitions.

**Result:** The extra transition variables did not consistently improve predictions. In the original path sample, 244 windows breached −5 pp and 149 finished below it.

For E18, `state_4_1.5` uses the last four weekly net returns divided by
$\sqrt{20}\sigma_t$. The state begins at a score ≤−1.5 and ends at a score ≥−.5;
between those boundaries it retains its previous state. `onset` predicts entry at
the next weekly cutoff among currently inactive rows; `exit` predicts departure
among active rows. `vol` uses log daily EWMA scale. `mechanical` adds the current
score and the oldest weekly return divided by the same scale, because that return
leaves the four-week window next week. `full` adds log state age, net beta,
rising/falling slope difference and the two averaged shape measures defined in the main
notebook. In `barrier_fixed`, the event instead is a daily-close loss ≤−5 pp during
the next twenty trading days; `full` there uses log scale, current stress score,
beta, slope difference and averaged shape measures, without state age. `fhs` is the
historical-path estimate. Fits are annual expanding standardised logits, C=.1, with
at least 100 completed labels, twenty events and twenty non-events. n counts eligible
cutoffs, not independent episodes. The plotted barrier sample starts 6 January 1984.

## E19. Path attribution, clustering and warning diagnostics

**Method:** I measured both legs at the first breach, subsequent low and endpoint, grouped overlapping windows, and examined false alarms, missed warnings and large-episode exclusions.

**Checks:** I reconciled both legs at each path stage and removed all windows intersecting an omitted overlap group.

**Result:** The short leg had the more negative first-breach contribution in 76% of breached windows. Weighting overlap groups equally changed the description of recovery contributions.

| Path stage, window-weighted mean | Long pp | Short pp | Net pp |
| --- | ---: | ---: | ---: |
| First breach | +.43 | −6.44 | −6.01 |
| Subsequent low | +.63 | −10.36 | −9.73 |
| Endpoint | +1.18 | −7.94 | −6.76 |

Equal-cluster first-breach contributions remain short-dominated (−.29/−5.47 pp),
but low-to-endpoint improvement becomes more balanced (+1.34/+1.16 pp). At an
illustrative 25% alert threshold, precision is 39.7% and window recall 50%. This
does not support a dependable alarm or a forecast of the next contributing leg.

<a id="e20"></a>
## E20. Frozen recent-period validation

**Method:** I extended the unchanged specifications through July 2026 and checked that earlier predictions reproduced the saved results.

**Checks:** I replayed earlier forecasts and measured annual scores, quantile breaches, scenario errors and path outcomes, without retuning.

**Result:** The recent historical-path estimates overpredict breaches: 19.59% estimated versus 9.29% observed. The weekly and scenario comparisons do not support adding the more complex predictors.

<a id="e21"></a>
## E21. GARCH simulations of interim loss

**Method:** I fitted annual daily Student-t GARCH and GJR models and simulated twenty-trading-day returns on the same dates and −5 pp target as the historical-path model.

**Checks:** I compared a constant-variance simulation, checked annual variance calculations and repeated recent simulations with more draws.

**Result:** GARCH has lower recent error, but still overpredicts recent breaches and has higher earlier log loss. The notebook displays both path estimates separately.

## Historical interpretation examples

The main notebook includes the weeks following 6 March 2009 and 13 March 2020.
The first illustrates rebound exposure with an elevated prior loss probability;
the second illustrates a loss whose prior estimated probability was below 20% and a conditional
response with the wrong net sign. The cases were selected retrospectively from saved
results. They are illustrations, not additional independent validation observations.
`scripts/macro_cases.py` reproduces the dated tables and figures and checks both legs
against the saved conditional predictions. Each example links a Federal Reserve
source published before its cutoff and separates subsequent outcomes from prior inputs.
No forecast or model parameter was changed to construct these cases.

## Reproduction and completeness

This appendix covers every documented numerical experiment family; the optional
archive retains the saved figures, tables and calculation extracts. Machine-readable score files, protocols,
source code and hashes are supplied in the research package. The main and compact
appendix can be read without that package; rerunning calculations requires it.
The archive is documentation of saved execution, not a new simultaneous refit.
Actual PM holdings, a continuous text archive, licensed positioning data and prospective
PM evaluation remain outside this prototype.
